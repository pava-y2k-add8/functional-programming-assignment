```python
#Q1. [Python Basics & Data Structures] Let A = {1, 2, 3, 4} and B = {3, 4, 5, 6}. Print A | B, A & B, A - B and A ^ B,each with a label. [5 Marks]

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(f"Intersection: {A|B}")
print(f"Union: {A&B}")
print(f"Difference: {A-B}")
print(f"Symmetric Difference: {A^B}")
```

```python
#Q2. [Strings, I/O & Operators] Let s = "data science". Print the position of "science" using find(), the result of find("x"),and whether the string starts with "data" and ends with "ce". [5 Marks]

s = "data science"

position_of_science = s.find("science")
print(f"The position of 'science': {position_of_science}")

result_of_find_x = s.find("x")
print(f"The result of find('x'): {result_of_find_x}")

starts_with_data = s.startswith("data")
print(f"String starts with 'data'?: {starts_with_data}")

ends_with_ce = s.endswith("ce")
print(f"String ends with 'ce'?: {ends_with_ce}")
```

```python
#Q3. [Control Flow & Loops] Repeatedly accept numbers from the user using a while loop until the user enters 0. Then print how many numbers were entered (excluding 0) and their sum. [5 Marks]

i = int(input("User please enter your number: "))
frequency = 0
summ = i

while i != 0:
  i = int(input("User please enter your number: "))
  frequency += 1
  summ = summ + i

print(f"The number of times numbers were entered is {frequency}")
print(f"The sum of the numbers entered is {summ}")
```
