```python
#Q1 [Data Types & Operators] Take a decimal value typed as a string, e.g. "3.9". Convert it directly to int() and observe/print the error message; then fix it by converting via float() first and then int(). [5 Marks]
print(int("3.9"))
print("ValueError: invalid literal for int() with base 10: '3.9'")
print("the answer is", int(float("3.9")))
```

```python
#Q2 [Strings] Write a program to remove ALL spaces from a sentence (not just the leading/trailing ones) using replace(). [5 Marks]
sentence = "My name is paavani maheshwari and I study at NMIMS."
rep = sentence.replace(" ","")
print(rep)
```

```python
#Q3. [Lists & Tuples] Take 5 numbers as a tuple and compute their AVERAGE without using the built-in sum() function. [5 Marks]
t = (1,2,3,4,5)
s = 0
for i in range(6):
    s = s + i 
print("the sum of all 5 numbers is",s)
print()
n = len(t)
print("the number of elements is",n)
print()
avg = int(s/n)
print("the average is",avg)

```
